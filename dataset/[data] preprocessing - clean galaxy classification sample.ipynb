## ------------ Classification Thresholds ------------

We classify the galaxies into 8 classes: round elliptical, in-between elliptical, cigar-shaped elliptical, edge-on, barred spiral, unbarred spiral, irregular, merger.

Threshold: following Table3 of [Willett+13](https://arxiv.org/pdf/1308.3496v2.pdf) or Table2 of [Kalvankar+20](https://arxiv.org/abs/2008.13611).

### 分类阈值相关说明
我们将星系分为 8 类：圆形椭圆星系、过渡型椭圆星系、雪茄形椭圆星系、侧视星系、棒旋星系、无棒旋星系、不规则星系、并合星系。

0∼7 in the order of: round elliptical, in-between elliptical, cigar-shaped elliptical, edge-on, barred spiral, unbarred spiral, irregular and merger galaxies.

阈值：参照威尔lett 等人 2013 年论文中的表 3 或卡尔万卡尔等人 2020 年论文中的表 2。 

# -----------images的图像处理-----------------

```python
import torch
from torchvision import transforms
from torch.utils.data import Dataset, DataLoader
import os
from PIL import Image

class AstroDataset(Dataset):
    def __init__(self, img_dir, text_labels, transform=None):
        self.img_dir = img_dir
        self.text_labels = text_labels  # 假设为包含图文对的字典 {galaxyID: "text description"}
        self.transform = transform
        self.img_ids = list(text_labels.keys())

    def __len__(self):
        return len(self.img_ids)

    def __getitem__(self, idx):
        img_id = self.img_ids[idx]
        img_path = os.path.join(self.img_dir, f"{img_id}.jpg")
        image = Image.open(img_path).convert("RGB")
        text = self.text_labels[img_id]
        
        if self.transform:
            image = self.transform(image)
        
        return image, text

# 示例：Qwen-VL 的预处理流程（假设预训练使用384x384和ImageNet统计量）
def get_transforms(target_size=384):
    # 训练集增强
    train_transform = transforms.Compose([
        transforms.RandomResizedCrop(
            target_size,
            scale=(0.7, 1.0),  # 扩大裁剪范围以适应天体位置变化
            ratio=(0.9, 1.1),  # 略微放宽宽高比限制
            interpolation=transforms.InterpolationMode.BICUBIC
        ),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(30),  # 限制旋转角度避免过度变形
        transforms.ToTensor(),
        transforms.Normalize(
            mean=[0.485, 0.456, 0.406],  # 沿用预训练统计量
            std=[0.229, 0.224, 0.225]
        )
    ])
    
    # 验证/测试集处理
    eval_transform = transforms.Compose([
        transforms.Resize(target_size + 32),  # 先缩放到稍大尺寸
        transforms.CenterCrop(target_size),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])
    
    return train_transform, eval_transform

# 初始化数据集和加载器
img_dir = "/remote-home/cs_acmis_hby/AstroData/images_gz2"
text_labels = {
    "8": "A round elliptical galaxy with smooth light distribution",
    "123": "A barred spiral galaxy with prominent central bar",
    # ... 其他图文对
}

train_transform, eval_transform = get_transforms(target_size=384)
train_dataset = AstroDataset(img_dir, text_labels, transform=train_transform)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=4, pin_memory=True)
```

# ------------GZ2图像路径与参数数据处理------------------

```python
import pandas as pd
import json
import os

def convert_class_files():
    # 配置路径和参数
    data_dir = "/remote-home/cs_acmis_hby/AstroData"
    input_file = os.path.join(data_dir, "classAll_4label.csv")  # 修改为新的输入文件
    image_base = f"{data_dir}/images_gz2"
    output_file = "/remote-home/cs_acmis_hby/code/FT-LLM/galaxy_classification/galaxy_C.json"
    
    # 定义星系类型映射字典
    type_map = {
        0: "round elliptical",
        1: "in-between elliptical",
        2: "cigar-shaped elliptical",
        3: "edge-on",
        4: "Barred spirals",
        5: "Unbarred spirals",
        6: "Irregular",
        7: "merger"
    }
    
    all_data = []
    conversion_stats = {}

    try:
        # 读取合并后的CSV文件
        df = pd.read_csv(input_file)
        print(f"成功读取主文件，总数据量: {len(df)} 行")

        # 按label1进行分类处理
        for class_num in type_map.keys():
            class_type = type_map[class_num]
            
            # 筛选当前类别的数据
            class_df = df[df['label1'] == class_num]
            count = len(class_df)
            
            print(f"\n▶ 正在处理 label1={class_num} ({class_type})")
            print(f"  匹配数据量: {count} 行")

            # 转换函数（保持原有逻辑）
            def convert_row(row):
                return {
                    "id": str(row["galaxyID"]),
                    "image": [f"{image_base}/{row['galaxyID']}.jpg"],
                    "conversations": [
                        {
                            "from": "human",
                            "value": "<image>\nGiven astronomical galaxy image, identify the galaxy morphological type based on the image.Return the answer in an Assignment Statement, containing ONE variable: IDENTIFICATION. Only return the IDENTIFICATION, not the Description"
                        },
                        {
                            "from": "gpt",
                            "value": f"IDENTIFICATION = '{class_type}'"
                        }
                    ]
                }

            # 执行转换
            class_data = class_df.apply(convert_row, axis=1).tolist()
            success_count = len(class_data)
            
            # 记录统计信息
            all_data.extend(class_data)
            conversion_stats[class_num] = {
                "count": success_count,
                "type": class_type
            }
            
            print(f"  ✓ 成功转换: {success_count} 条")
            print(f"  ░ 累计总数: {len(all_data)} 条")

    except Exception as e:
        print(f"✕ 发生致命错误: {str(e)}")
        return

    # 保存最终文件
    with open(output_file, 'w') as f:
        json.dump(all_data, f, indent=2)
    
    # 生成统计报告
    print("\n▄▄▄▄▄▄▄▄▄▄ 转换统计报告 ▄▄▄▄▄▄▄▄▄▄")
    total = 0
    for k, v in conversion_stats.items():
        print(f"label1={k} ({type_map[k].ljust(22)}): {str(v['count']).rjust(5)} 条")
        total += v['count']
    print("▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄▄")
    print(f"总转换数量: {total} 条")
    print(f"输出文件路径: {os.path.abspath(output_file)}")

# 执行转换
convert_class_files()
```

# 处理生成星系带参数的json文件

```python
import pandas as pd
import json
import os

def convert_class_files_with_features():
    # ================= 配置参数 =================
    DATA_DIR = "/remote-home/cs_acmis_hby/AstroData"
    INPUT_CSV = os.path.join(DATA_DIR, "parameter_features.csv")
    IMAGE_BASE = f"{DATA_DIR}/images_gz2"
    OUTPUT_JSON = "/remote-home/cs_acmis_hby/code/FT-LLM/galaxy_classification/galaxy_Para_with_features.json"
    
    # ================= 类型映射 =================
    GALAXY_TYPES = {
        0: ("round elliptical", "A"),
        1: ("in-between elliptical", "B"),
        2: ("cigar-shaped elliptical", "C"),
        3: ("edge-on", "D"),
        4: ("Barred spirals", "E"),
        5: ("Unbarred spirals", "F"),
        6: ("Irregular", "G"),
        7: ("merger", "H")
    }
    
    # ================= 特征列映射 =================
    FEATURE_MAPPING = {
        't01_1': 'f_smooth',
        't01_2': 'f_features/disk',
        't02_1': 'f_edge-on/yes',
        't02_2': 'f_edge-on/no',
        't03_1': 'f_bar/yes',
        't03_2': 'f_bar/no',
        't04_1': 'f_spiral/yes',
        't07_1': 'f_completelyround',
        't07_2': 'f_in-between',
        't07_3': 'f_cigar-shaped',
        't06_1': 'f_odd/yes',
        't06_2': 'f_odd/no',
        't08_3': 'f_disturbed',
        't08_4': 'f_irregular',
        't08_5': 'f_other',
        't08_6': 'f_merger',
        't08_7': 'f_dustlane'
    }

    # ================= 分类阈值标准 =================
    CLASS_CRITERIA = {
        0: ["f_smooth >= 0.469", "f_completelyround >= 0.5", "f_odd/no >= 0.5"],
        1: ["f_smooth >= 0.469", "f_in-between >= 0.5", "f_odd/no >= 0.5"],
        2: ["f_smooth >= 0.469", "f_cigar-shaped >= 0.5", "f_odd/no >= 0.5"],
        3: ["f_features/disk >= 0.430", "f_edge-on/yes >= 0.602", "f_odd/no >= 0.5"],
        4: ["f_features/disk >= 0.430", "f_edge-on/no >= 0.715", "f_bar/yes >= 0.715", "f_spiral/yes >= 0.619"],
        5: ["f_features/disk >= 0.430", "f_edge-on/no >= 0.715", "f_bar/no >= 0.715", "f_spiral/yes >= 0.619"],
        6: ["f_odd/yes >= 0.420", "(f_disturbed >= 0.5 OR f_irregular >= 0.5 OR f_other >= 0.5 OR f_dustlane >= 0.5)"],
        7: ["f_odd/yes >= 0.420", "f_merger >= 0.5"]
    }

    # ================= 主处理逻辑 =================
    try:
        # 读取数据并重命名列
        df = pd.read_csv(
            INPUT_CSV,
            dtype={'galaxyID': 'Int64'},
            usecols=list(FEATURE_MAPPING.keys()) + ['galaxyID', 'label1']
        ).rename(columns=FEATURE_MAPPING)

        print(f"✅ 成功读取数据文件，总样本数: {len(df):,}")

        # 校验数据完整性
        required_cols = list(FEATURE_MAPPING.values()) + ['galaxyID', 'label1']
        if missing := [col for col in required_cols if col not in df.columns]:
            raise KeyError(f"缺失关键列: {missing}")

        # 清洗galaxyID
        if not df['galaxyID'].apply(lambda x: x == int(x)).all():
            invalid = df[df['galaxyID'] != df['galaxyID'].astype(int)]['galaxyID']
            print(f"⚠️ 警告: 发现{len(invalid)}个非整型ID，执行取整操作")
            df['galaxyID'] = df['galaxyID'].astype(int)

        # 初始化数据结构
        dataset = []
        stats = {k: {"count":0, "type":v[0]} for k,v in GALAXY_TYPES.items()}

        for class_id, (class_name, option_char) in GALAXY_TYPES.items():
            class_df = df[df['label1'] == class_id]
            if class_df.empty:
                print(f"⚠️ 警告: 类别{class_id}({class_name})无数据")
                continue

            def create_conversation(row):
                # 生成特征字符串
                features = ",\n".join([f"{k}={v:.5f}" for k,v in row.items() if k in FEATURE_MAPPING.values()])
                
                # 生成选项列表（严格遵循格式）
                options = ",\n".join([f"'{v[1]}:{v[0]}'" for v in GALAXY_TYPES.values()])
                
                # 获取满足的条件
                met_conditions = []
                for condition in CLASS_CRITERIA[class_id]:
                    if "OR" in condition:
                        parts = [c.strip() for c in condition.strip("()").split("OR")]
                        for part in parts:
                            var, val = part.split(">=")
                            if row[var.strip()] >= float(val.strip()):
                                met_conditions.append(condition.replace(" OR ", " or "))
                                break
                    else:
                        var, val = condition.split(">=")
                        if row[var.strip()] >= float(val.strip()):
                            met_conditions.append(condition)

                return {
                    "id": int(row["galaxyID"]),
                    "image": [f"{IMAGE_BASE}/{int(row['galaxyID'])}.jpg"],
                    "conversations": [
                        {
                            "from": "human",
                            "value": (
                                f"<image>Given galaxy image and the corresponding physical parameters:\n"
                                f"{features}\n\n"
                                f"[Output Constraints] Return the answer as one of the following options:\n"
                                f"{options}\n\n"
                                "Now, based on the morphological features and image, return the choice."
                            )
                        },
                        {
                            "from": "gpt",
                            "value": (
                                "Answer:\n"
                                "Based on the image information, and the corresponding feature parameters of the image meet the following conditions:\n"
                                f"{'; '.join(met_conditions)}\n"
                                f"Therefore, option {option_char} is selected."
                            )
                        }
                    ]
                }

            converted = class_df.apply(create_conversation, axis=1).tolist()
            stats[class_id]["count"] = len(converted)
            dataset.extend(converted)
            print(f"🔄 转换完成 {class_name.ljust(22)}: {len(converted)} 条")

        # 保存结果
        with open(OUTPUT_JSON, 'w') as f:
            json.dump(dataset, f, indent=2, ensure_ascii=False)

        # 生成报告
        print("\n▄▄▄▄▄▄▄▄▄ 转换统计 ▄▄▄▄▄▄▄▄▄")
        total = sum(info["count"] for info in stats.values())
        for k in sorted(stats.keys()):
            print(f"类别 {k}: {stats[k]['type'].ljust(24)} {stats[k]['count']:>5} 条")
        print(f"总转换数量: {total} 条")
        print(f"输出文件: {os.path.abspath(OUTPUT_JSON)}")

    except Exception as e:
        print(f"❌ 严重错误: {type(e).__name__} - {str(e)}")

if __name__ == "__main__":
    convert_class_files_with_features()
```

```python
import json
import os
import re
import numpy as np
from sklearn.model_selection import train_test_split
from collections import defaultdict, Counter
from PIL import Image

def process_galaxy_data():
    # ---------------------- 配置路径 ----------------------
    json_path = "/remote-home/cs_acmis_hby/code/FT-LLM/galaxy_classification/galaxy_Para_with_features_02.json"
    original_img_dir = "/remote-home/cs_acmis_hby/AstroData/images_gz2/images"
    output_dir = "/remote-home/cs_acmis_hby/code/FT-LLM/galaxy_classification/dataset_07"
    
    # 创建输出目录
    train_img_dir = os.path.join(output_dir, "train_images")
    test_img_dir = os.path.join(output_dir, "test_images")
    os.makedirs(train_img_dir, exist_ok=True)
    os.makedirs(test_img_dir, exist_ok=True)

    # ---------------------- 数据加载与清洗 ----------------------
    print("⏳ 正在加载原始数据...")
    try:
        with open(json_path, 'r') as f:
            raw_data = json.load(f)
        print(f"✅ 成功加载 {len(raw_data)} 条原始数据")
    except Exception as e:
        raise RuntimeError(f"数据加载失败: {str(e)}")

    # 改进的标签提取正则表达式
    LABEL_PATTERN = re.compile(r"option\s*([A-H])", re.IGNORECASE)
    
    def extract_label(content):
        """改进的标签解析函数"""
        match = LABEL_PATTERN.search(content)
        if match:
            return match.group(1).upper()
        # 备用解析逻辑
        if "option A" in content: return "A"
        if "option B" in content: return "B"
        return None

    # 构建格式化数据
    formatted_data = []
    label_counter = Counter()
    missing_images = 0
    label_parse_errors = 0

    print("\n🔍 正在进行数据清洗...")
    for idx, item in enumerate(raw_data):
        try:
            # 验证图像存在性
            valid_images = []
            for img_rel_path in item["image"]:
                img_name = os.path.basename(img_rel_path)
                src_path = os.path.join(original_img_dir, img_name)
                
                if os.path.exists(src_path):
                    valid_images.append(img_name)
                else:
                    missing_images += 1
                    if idx < 5:  # 打印前5个缺失样本
                        print(f"  警告：图像缺失 {img_name}")

            if not valid_images:
                continue

            # 构建对话消息
            messages = []
            for conv in item["conversations"]:
                role = "user" if conv["from"] == "human" else "assistant"
                content = conv["value"]
                
                if role == "user":
                    content = "<image>" + content.split("\n", 1)[-1].strip()
                
                messages.append({
                    "role": role,
                    "content": content.strip()
                })

            # 提取标签
            assistant_content = next(msg["content"] for msg in messages if msg["role"] == "assistant")
            label = extract_label(assistant_content)
            
            if not label:
                label_parse_errors += 1
                if label_parse_errors <= 3:  # 打印前3个解析错误
                    print(f"  标签解析失败: {assistant_content[:50]}...")
                continue
                
            label_counter.update([label])
            formatted_data.append({
                "messages": messages,
                "images": valid_images.copy()
            })

        except Exception as e:
            print(f"⚠️ 数据处理异常（条目{idx}）: {str(e)}")
            continue

    # ---------------------- 数据验证 ----------------------
    print("\n📊 数据清洗结果:")
    print(f" - 有效数据条目: {len(formatted_data)}")
    print(f" - 缺失图像数: {missing_images}")
    print(f" - 标签解析失败数: {label_parse_errors}")
    print(" - 类别分布:")
    for label, count in label_counter.most_common():
        print(f"   {label}: {count} 样本")

    if not label_counter:
        raise ValueError("⚠️ 所有标签解析失败，请检查数据格式！")

    # ---------------------- 分层抽样 ----------------------
    min_samples = 1  # 允许单个样本类别
    valid_labels = [label for label, count in label_counter.items() if count >= min_samples]
    
    print(f"\n🔖 有效类别（≥{min_samples}样本）: {valid_labels}")
    if not valid_labels:
        raise ValueError(f"⚠️ 没有满足最小样本数的类别（当前min_samples={min_samples}）")

    # 构建有效数据索引
    valid_data = []
    label_indices = defaultdict(list)
    for idx, item in enumerate(formatted_data):
        assistant_content = next(msg["content"] for msg in item["messages"] if msg["role"] == "assistant")
        if (label := extract_label(assistant_content)) in valid_labels:
            label_indices[label].append(idx)
            valid_data.append(item)

    # 执行安全抽样
    sampled_indices = []
    print("\n🎲 正在进行分层抽样...")
    for label, indices in label_indices.items():
        sample_size = min(len(indices), 8000)
        if len(indices) < min_samples:
            print(f"  警告：类别 {label} 样本不足（{len(indices)}），已跳过")
            continue
            
        print(f"  {label}: 抽样 {sample_size}/{len(indices)}")
        sampled = np.random.choice(indices, size=sample_size, replace=False)
        sampled_indices.extend(sampled.tolist())
    
    sampled_data = [valid_data[i] for i in sampled_indices]
    print(f"\n📦 抽样后数据量: {len(sampled_data)}")

    # ---------------------- 数据集划分 ----------------------
    print("\n✂️ 正在划分数据集...")
    labels = [extract_label(next(msg["content"] for msg in item["messages"] if msg["role"] == "assistant")) 
             for item in sampled_data]

    # 动态调整参数
    test_size = 0.2 if len(sampled_data) >= 10 else 0.1
    stratify = labels if len(set(labels)) > 1 else None
    
    try:
        train_data, test_data = train_test_split(
            sampled_data,
            test_size=test_size,
            stratify=stratify,
            random_state=42
        )
    except ValueError as e:
        print(f"⚠️ 分层划分失败: {str(e)}")
        print("  尝试非分层划分...")
        train_data, test_data = train_test_split(
            sampled_data,
            test_size=test_size,
            stratify=None,
            random_state=42
        )

    print(f" 训练集: {len(train_data)} | 测试集: {len(test_data)}")

    # ---------------------- 图像处理 ----------------------
    def process_and_copy_images(data, target_img_dir):
        """安全图像处理器"""
        processed_data = []
        for item in data:
            new_images = []
            for img_name in item["images"]:
                src_path = os.path.join(original_img_dir, img_name)
                dst_path = os.path.join(target_img_dir, img_name)
                
                try:
                    with Image.open(src_path) as img:
                        # 中心裁剪
                        w, h = img.size
                        crop_box = (
                            (w - 224) // 2,
                            (h - 224) // 2,
                            (w + 224) // 2,
                            (h + 224) // 2
                        )
                        cropped = img.crop(crop_box)
                        
                        if cropped.mode != 'RGB':
                            cropped = cropped.convert('RGB')
                            
                        cropped.save(dst_path)
                        new_images.append(dst_path)
                except Exception as e:
                    print(f"⚠️ 图像处理失败: {img_name} -> {str(e)}")
                    continue
            
            if new_images:
                processed_data.append({
                    "messages": [
                        {
                            "role": msg["role"],
                            "content": msg["content"].replace("\n", " ").strip()
                        } for msg in item["messages"]
                    ],
                    "images": new_images
                })
        return processed_data

    print("\n🖼️ 正在处理图像...")
    print("  训练集图像处理中...")
    train_processed = process_and_copy_images(train_data, train_img_dir)
    print("  测试集图像处理中...")
    test_processed = process_and_copy_images(test_data, test_img_dir)

    # ---------------------- 数据保存 ----------------------
    def save_jsonl(data, path):
        with open(path, 'w') as f:
            for item in data:
                # 宽松格式验证
                try:
                    assert len(item["messages"]) >= 1, "消息数量不足"
                    assert any(msg["role"] == "user" for msg in item["messages"]), "缺少用户消息"
                    assert any(msg["role"] == "assistant" for msg in item["messages"]), "缺少助手消息"
                except AssertionError as e:
                    print(f"⚠️ 数据格式异常: {e}")
                    continue
                    
                f.write(json.dumps(item, ensure_ascii=False) + '\n')

    print("\n💾 正在保存数据集...")
    save_jsonl(train_processed, os.path.join(output_dir, "train.jsonl"))
    save_jsonl(test_processed, os.path.join(output_dir, "test.jsonl"))

    # ---------------------- 最终报告 ----------------------
    print("\n" + "="*40)
    print("🏁 处理完成！最终统计:")
    print(f" - 原始数据量: {len(raw_data)}")
    print(f" - 有效数据量: {len(formatted_data)}")
    print(f" - 训练集数量: {len(train_processed)}")
    print(f" - 测试集数量: {len(test_processed)}")
    
    if train_processed:
        sample = train_processed[0]
        print("\n🔍 示例样本:")
        print("用户消息:", sample["messages"][0]["content"][:80] + "...")
        print("助手消息:", sample["messages"][1]["content"][:80] + "...")
        print("图像路径:", sample["images"][0])
    else:
        print("\n⚠️ 警告：训练集为空！")

if __name__ == "__main__":
    process_galaxy_data()
```

```python
import pandas as pd
import json
import os

def convert_class_files_with_features():
    # ================= 配置参数 =================
    DATA_DIR = "/remote-home/cs_acmis_hby/AstroData"
    INPUT_CSV = os.path.join(DATA_DIR, "parameter_features.csv")
    IMAGE_BASE = f"{DATA_DIR}/images_gz2/images"
    OUTPUT_JSON = "/remote-home/cs_acmis_hby/code/FT-LLM/galaxy_classification/galaxy_Para_with_features_04.json"
    
    # ================= 类型映射 =================
    GALAXY_TYPES = {
        0: ("round elliptical", "A"),
        1: ("in-between elliptical", "B"),
        2: ("cigar-shaped elliptical", "C"),
        3: ("edge-on", "D"),
        4: ("Barred spirals", "E"),
        5: ("Unbarred spirals", "F"),
        6: ("Irregular", "G"),
        7: ("merger", "H")
    }
    
    # ================= 特征列映射 =================
    FEATURE_MAPPING = {
        't01_1': 'f_smooth',
        't01_2': 'f_features/disk',
        't02_1': 'f_edge-on/yes',
        't02_2': 'f_edge-on/no',
        't03_1': 'f_bar/yes',
        't03_2': 'f_bar/no',
        't04_1': 'f_spiral/yes',
        't07_1': 'f_completelyround',
        't07_2': 'f_in-between',
        't07_3': 'f_cigar-shaped',
        't06_1': 'f_odd/yes',
        't06_2': 'f_odd/no',
        't08_3': 'f_disturbed',
        't08_4': 'f_irregular',
        't08_5': 'f_other',
        't08_6': 'f_merger',
        't08_7': 'f_dustlane'
    }

    # ================= 分类阈值标准 =================
    CLASS_CRITERIA = {
        0: ["f_smooth >= 0.469", "f_completelyround >= 0.5", "f_odd/no >= 0.5"],
        1: ["f_smooth >= 0.469", "f_in-between >= 0.5", "f_odd/no >= 0.5"],
        2: ["f_smooth >= 0.469", "f_cigar-shaped >= 0.5", "f_odd/no >= 0.5"],
        3: ["f_features/disk >= 0.430", "f_edge-on/yes >= 0.602", "f_odd/no >= 0.5"],
        4: ["f_features/disk >= 0.430", "f_edge-on/no >= 0.715", "f_bar/yes >= 0.715", "f_spiral/yes >= 0.619"],
        5: ["f_features/disk >= 0.430", "f_edge-on/no >= 0.715", "f_bar/no >= 0.715", "f_spiral/yes >= 0.619"],
        6: ["f_odd/yes >= 0.420", "(f_disturbed >= 0.5 OR f_irregular >= 0.5 OR f_other >= 0.5 OR f_dustlane >= 0.5)"],
        7: ["f_odd/yes >= 0.420", "f_merger >= 0.5"]
    }

    # ================= 主处理逻辑 =================
    try:
        # 读取数据并重命名列
        df = pd.read_csv(
            INPUT_CSV,
            dtype={'galaxyID': 'Int64'},
            usecols=list(FEATURE_MAPPING.keys()) + ['galaxyID', 'label1']
        ).rename(columns=FEATURE_MAPPING)

        print(f"✅ 成功读取数据文件，总样本数: {len(df):,}")

        # 校验数据完整性
        required_cols = list(FEATURE_MAPPING.values()) + ['galaxyID', 'label1']
        if missing := [col for col in required_cols if col not in df.columns]:
            raise KeyError(f"缺失关键列: {missing}")

        # 清洗galaxyID
        if not df['galaxyID'].apply(lambda x: x == int(x)).all():
            invalid = df[df['galaxyID'] != df['galaxyID'].astype(int)]['galaxyID']
            print(f"⚠️ 警告: 发现{len(invalid)}个非整型ID，执行取整操作")
            df['galaxyID'] = df['galaxyID'].astype(int)

        # 初始化数据结构
        dataset = []
        stats = {k: {"count":0, "type":v[0]} for k,v in GALAXY_TYPES.items()}

        for class_id, (class_name, option_char) in GALAXY_TYPES.items():
            class_df = df[df['label1'] == class_id]
            if class_df.empty:
                print(f"⚠️ 警告: 类别{class_id}({class_name})无数据")
                continue

            def create_conversation(row):
                # 生成特征字符串
                features = ",\n".join([f"{k}={v:.5f}" for k,v in row.items() if k in FEATURE_MAPPING.values()])
                
                # 生成选项列表（严格遵循格式）
                options = ",\n".join([f"'{v[1]}:{v[0]}'" for v in GALAXY_TYPES.values()])
                
                # 获取满足的条件
                met_conditions = []
                for condition in CLASS_CRITERIA[class_id]:
                    if "OR" in condition:
                        parts = [c.strip() for c in condition.strip("()").split("OR")]
                        for part in parts:
                            var, val = part.split(">=")
                            if row[var.strip()] >= float(val.strip()):
                                met_conditions.append(condition.replace(" OR ", " or "))
                                break
                    else:
                        var, val = condition.split(">=")
                        if row[var.strip()] >= float(val.strip()):
                            met_conditions.append(condition)

                return {
                    "id": int(row["galaxyID"]),
                    "image": [f"{IMAGE_BASE}/{int(row['galaxyID'])}.jpg"],
                    "conversations": [
                        {
                            "from": "human",
                            "value": (
                                f"<image>Given galaxy image:\n"
                                f"{options}\n"
                                f"Galaxy morphology is option {option_char}.\n"
                                "Now, based on the choice and image, return the features of the galaxy  morphology."
                            )
                        },
                        {
                            "from": "gpt",
                            "value": (
                                "Answer:\n"
                                f"Based on the image information and the option {option_char} is selected.\n"
                                f"Therefore, the corresponding feature parameters of the image are:\n"
                                f"{features}\n\n"  
                            )
                        }
                    ]
                }

            converted = class_df.apply(create_conversation, axis=1).tolist()
            stats[class_id]["count"] = len(converted)
            dataset.extend(converted)
            print(f"🔄 转换完成 {class_name.ljust(22)}: {len(converted)} 条")

        # 保存结果
        with open(OUTPUT_JSON, 'w') as f:
            json.dump(dataset, f, indent=2, ensure_ascii=False)

        # 生成报告
        print("\n▄▄▄▄▄▄▄▄▄ 转换统计 ▄▄▄▄▄▄▄▄▄")
        total = sum(info["count"] for info in stats.values())
        for k in sorted(stats.keys()):
            print(f"类别 {k}: {stats[k]['type'].ljust(24)} {stats[k]['count']:>5} 条")
        print(f"总转换数量: {total} 条")
        print(f"输出文件: {os.path.abspath(OUTPUT_JSON)}")

    except Exception as e:
        print(f"❌ 严重错误: {type(e).__name__} - {str(e)}")

if __name__ == "__main__":
    convert_class_files_with_features()
```

```python
import json
import os
import re
import numpy as np
from sklearn.model_selection import train_test_split
from collections import defaultdict, Counter
from PIL import Image

def process_galaxy_data():
    # ---------------------- 配置路径 ----------------------
    json_path = "/remote-home/cs_acmis_hby/code/FT-LLM/galaxy_classification/galaxy_Para_with_features_04.json"
    original_img_dir = "/remote-home/cs_acmis_hby/AstroData/images_gz2/images"
    output_dir = "/remote-home/cs_acmis_hby/code/FT-LLM/galaxy_classification/dataset_05"
    
    # 创建输出目录
    train_img_dir = os.path.join(output_dir, "train_images")
    test_img_dir = os.path.join(output_dir, "test_images")
    os.makedirs(train_img_dir, exist_ok=True)
    os.makedirs(test_img_dir, exist_ok=True)

    # ---------------------- 数据加载与清洗 ----------------------
    print("⏳ 正在加载原始数据...")
    try:
        with open(json_path, 'r') as f:
            raw_data = json.load(f)
        print(f"✅ 成功加载 {len(raw_data)} 条原始数据")
    except Exception as e:
        raise RuntimeError(f"数据加载失败: {str(e)}")

    # 改进的标签提取正则表达式
    LABEL_PATTERN = re.compile(r"option\s*([A-H])", re.IGNORECASE)
    
    def extract_label(content):
        """改进的标签解析函数"""
        match = LABEL_PATTERN.search(content)
        if match:
            return match.group(1).upper()
        # 备用解析逻辑
        if "option A" in content: return "A"
        if "option B" in content: return "B"
        return None

    # 构建格式化数据
    formatted_data = []
    label_counter = Counter()
    missing_images = 0
    label_parse_errors = 0

    print("\n🔍 正在进行数据清洗...")
    for idx, item in enumerate(raw_data):
        try:
            # 验证图像存在性
            valid_images = []
            for img_rel_path in item["image"]:
                img_name = os.path.basename(img_rel_path)
                src_path = os.path.join(original_img_dir, img_name)
                
                if os.path.exists(src_path):
                    valid_images.append(img_name)
                else:
                    missing_images += 1
                    if idx < 5:  # 打印前5个缺失样本
                        print(f"  警告：图像缺失 {img_name}")

            if not valid_images:
                continue

            # 构建对话消息
            messages = []
            for conv in item["conversations"]:
                role = "user" if conv["from"] == "human" else "assistant"
                content = conv["value"]
                
                if role == "user":
                    content = "<image>" + content.split("\n", 1)[-1].strip()
                
                messages.append({
                    "role": role,
                    "content": content.strip()
                })

            # 提取标签
            assistant_content = next(msg["content"] for msg in messages if msg["role"] == "assistant")
            label = extract_label(assistant_content)
            
            if not label:
                label_parse_errors += 1
                if label_parse_errors <= 3:  # 打印前3个解析错误
                    print(f"  标签解析失败: {assistant_content[:50]}...")
                continue
                
            label_counter.update([label])
            formatted_data.append({
                "messages": messages,
                "images": valid_images.copy()
            })

        except Exception as e:
            print(f"⚠️ 数据处理异常（条目{idx}）: {str(e)}")
            continue

    # ---------------------- 数据验证 ----------------------
    print("\n📊 数据清洗结果:")
    print(f" - 有效数据条目: {len(formatted_data)}")
    print(f" - 缺失图像数: {missing_images}")
    print(f" - 标签解析失败数: {label_parse_errors}")
    print(" - 类别分布:")
    for label, count in label_counter.most_common():
        print(f"   {label}: {count} 样本")

    if not label_counter:
        raise ValueError("⚠️ 所有标签解析失败，请检查数据格式！")

    # ---------------------- 分层抽样 ----------------------
    min_samples = 1  # 允许单个样本类别
    valid_labels = [label for label, count in label_counter.items() if count >= min_samples]
    
    print(f"\n🔖 有效类别（≥{min_samples}样本）: {valid_labels}")
    if not valid_labels:
        raise ValueError(f"⚠️ 没有满足最小样本数的类别（当前min_samples={min_samples}）")

    # 构建有效数据索引
    valid_data = []
    label_indices = defaultdict(list)
    for idx, item in enumerate(formatted_data):
        assistant_content = next(msg["content"] for msg in item["messages"] if msg["role"] == "assistant")
        if (label := extract_label(assistant_content)) in valid_labels:
            label_indices[label].append(idx)
            valid_data.append(item)

    # 执行安全抽样
    sampled_indices = []
    print("\n🎲 正在进行分层抽样...")
    for label, indices in label_indices.items():
        sample_size = min(len(indices), 4000)
        if len(indices) < min_samples:
            print(f"  警告：类别 {label} 样本不足（{len(indices)}），已跳过")
            continue
            
        print(f"  {label}: 抽样 {sample_size}/{len(indices)}")
        sampled = np.random.choice(indices, size=sample_size, replace=False)
        sampled_indices.extend(sampled.tolist())
    
    sampled_data = [valid_data[i] for i in sampled_indices]
    print(f"\n📦 抽样后数据量: {len(sampled_data)}")

    # ---------------------- 数据集划分 ----------------------
    print("\n✂️ 正在划分数据集...")
    labels = [extract_label(next(msg["content"] for msg in item["messages"] if msg["role"] == "assistant")) 
             for item in sampled_data]

    # 动态调整参数
    test_size = 0.2 if len(sampled_data) >= 10 else 0.1
    stratify = labels if len(set(labels)) > 1 else None
    
    try:
        train_data, test_data = train_test_split(
            sampled_data,
            test_size=test_size,
            stratify=stratify,
            random_state=42
        )
    except ValueError as e:
        print(f"⚠️ 分层划分失败: {str(e)}")
        print("  尝试非分层划分...")
        train_data, test_data = train_test_split(
            sampled_data,
            test_size=test_size,
            stratify=None,
            random_state=42
        )

    print(f" 训练集: {len(train_data)} | 测试集: {len(test_data)}")

    # ---------------------- 图像处理 ----------------------
    def process_and_copy_images(data, target_img_dir):
        """安全图像处理器"""
        processed_data = []
        for item in data:
            new_images = []
            for img_name in item["images"]:
                src_path = os.path.join(original_img_dir, img_name)
                dst_path = os.path.join(target_img_dir, img_name)
                
                try:
                    with Image.open(src_path) as img:
                        # 中心裁剪
                        w, h = img.size
                        crop_box = (
                            (w - 224) // 2,
                            (h - 224) // 2,
                            (w + 224) // 2,
                            (h + 224) // 2
                        )
                        cropped = img.crop(crop_box)
                        
                        if cropped.mode != 'RGB':
                            cropped = cropped.convert('RGB')
                            
                        cropped.save(dst_path)
                        new_images.append(dst_path)
                except Exception as e:
                    print(f"⚠️ 图像处理失败: {img_name} -> {str(e)}")
                    continue
            
            if new_images:
                processed_data.append({
                    "messages": [
                        {
                            "role": msg["role"],
                            "content": msg["content"].replace("\n", " ").strip()
                        } for msg in item["messages"]
                    ],
                    "images": new_images
                })
        return processed_data

    print("\n🖼️ 正在处理图像...")
    print("  训练集图像处理中...")
    train_processed = process_and_copy_images(train_data, train_img_dir)
    print("  测试集图像处理中...")
    test_processed = process_and_copy_images(test_data, test_img_dir)

    # ---------------------- 数据保存 ----------------------
    def save_jsonl(data, path):
        with open(path, 'w') as f:
            for item in data:
                # 宽松格式验证
                try:
                    assert len(item["messages"]) >= 1, "消息数量不足"
                    assert any(msg["role"] == "user" for msg in item["messages"]), "缺少用户消息"
                    assert any(msg["role"] == "assistant" for msg in item["messages"]), "缺少助手消息"
                except AssertionError as e:
                    print(f"⚠️ 数据格式异常: {e}")
                    continue
                    
                f.write(json.dumps(item, ensure_ascii=False) + '\n')

    print("\n💾 正在保存数据集...")
    save_jsonl(train_processed, os.path.join(output_dir, "train.jsonl"))
    save_jsonl(test_processed, os.path.join(output_dir, "test.jsonl"))

    # ---------------------- 最终报告 ----------------------
    print("\n" + "="*40)
    print("🏁 处理完成！最终统计:")
    print(f" - 原始数据量: {len(raw_data)}")
    print(f" - 有效数据量: {len(formatted_data)}")
    print(f" - 训练集数量: {len(train_processed)}")
    print(f" - 测试集数量: {len(test_processed)}")
    
    if train_processed:
        sample = train_processed[0]
        print("\n🔍 示例样本:")
        print("用户消息:", sample["messages"][0]["content"][:80] + "...")
        print("助手消息:", sample["messages"][1]["content"][:80] + "...")
        print("图像路径:", sample["images"][0])
    else:
        print("\n⚠️ 警告：训练集为空！")

if __name__ == "__main__":
    process_galaxy_data()
```

```python
import json
import os
import shutil
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from collections import defaultdict

def process_galaxy_data():
    # ---------------------- 配置路径 ----------------------
    json_path = "/remote-home/cs_acmis_hby/code/FT-LLM/galaxy_classification/galaxy_C.json"
    original_img_dir = "/remote-home/cs_acmis_hby/AstroData/images_gz2/images"
    output_dir = "/remote-home/cs_acmis_hby/code/FT-LLM/galaxy_classification/dataset"
    
    # 创建输出目录结构
    os.makedirs(os.path.join(output_dir, "train_images"), exist_ok=True)
    os.makedirs(os.path.join(output_dir, "test_images"), exist_ok=True)

    # ---------------------- 数据加载与清洗 ----------------------
    with open(json_path, 'r') as f:
        raw_data = json.load(f)
    
    # 数据清洗和格式转换
    formatted_data = []
    for item in raw_data:
        # 转换对话格式
        messages = []
        for conv in item["conversations"]:
            if conv["from"] == "human":
                role = "user"
            elif conv["from"] == "gpt":
                role = "assistant"
            else:
                continue  # 过滤未知角色
            
            # 处理多模态标记
            content = conv["value"]
            if "image" in item and len(item["image"]) > 0:
                content = " ".join([f"<image>{os.path.basename(img)}" 
                                  for img in item["image"]]) + " " + content
            
            messages.append({
                "role": role,
                "content": content.strip()
            })
        
        # 构建标准格式
        formatted_item = {
            "messages": messages,
            "images": [os.path.basename(img) for img in item["image"]]
        }
        formatted_data.append(formatted_item)

    # ---------------------- 分层抽样逻辑 ----------------------
    # 提取分类标签用于分层抽样
    label_dict = defaultdict(list)
    for idx, item in enumerate(formatted_data):
        for msg in item["messages"]:
            if msg["role"] == "assistant":
                label = msg["content"].split("is ")[-1].strip('.').lower()
                label_dict[label].append(idx)
                break

    # 执行分层抽样
    sampled_indices = []
    for label, indices in label_dict.items():
        sample_size = min(len(indices), 1000)
        sampled = np.random.choice(indices, size=sample_size, replace=False)
        sampled_indices.extend(sampled.tolist())
    
    sampled_data = [formatted_data[i] for i in sampled_indices]

    # ---------------------- 数据集划分 ----------------------
    # 提取标签用于分层
    labels = []
    for item in sampled_data:
        for msg in item["messages"]:
            if msg["role"] == "assistant":
                label = msg["content"].split("is ")[-1].strip('.').lower()
                labels.append(label)
                break

    # 分层划分
    train_data, test_data = train_test_split(
        sampled_data,
        test_size=0.2,
        stratify=labels,
        random_state=42
    )

    # ---------------------- 图像文件操作 ----------------------
    def copy_media_files(data, target_img_dir):
        """复制关联的媒体文件"""
        os.makedirs(target_img_dir, exist_ok=True)
        success = 0
        for item in data:
            for img in item["images"]:
                src_path = os.path.join(original_img_dir, img)
                dst_path = os.path.join(target_img_dir, img)
                try:
                    shutil.copy(src_path, dst_path)
                    success += 1
                except Exception as e:
                    print(f"Error copying {img}: {str(e)}")
        return success

    # 复制训练集和测试集文件
    train_img_dir = os.path.join(output_dir, "train_images")
    test_img_dir = os.path.join(output_dir, "test_images")
    train_copy_count = copy_media_files(train_data, train_img_dir)
    test_copy_count = copy_media_files(test_data, test_img_dir)

    # ---------------------- 路径标准化 ----------------------
    def update_paths(data, img_folder):
        """更新图像路径为相对路径"""
        for item in data:
            item["images"] = [os.path.join(img_folder, img) for img in item["images"]]
        return data

    train_data = update_paths(train_data, "train_images")
    test_data = update_paths(test_data, "test_images")

    # ---------------------- 数据保存 ----------------------
    # 保存为JSONL格式（每行一个JSON对象）
    def save_jsonl(data, path):
        with open(path, 'w') as f:
            for item in data:
                f.write(json.dumps(item, ensure_ascii=False) + '\n')

    train_path = os.path.join(output_dir, "train.jsonl")
    test_path = os.path.join(output_dir, "test.jsonl")
    save_jsonl(train_data, train_path)
    save_jsonl(test_data, test_path)

    # ---------------------- 统计报告 ----------------------
    print("\n▣▣▣ 数据处理结果报告 ▣▣▣")
    print(f"原始数据总量: {len(raw_data)}")
    print(f"有效样本数量: {len(sampled_data)}")
    print(f"训练集数量: {len(train_data)} (图片复制成功: {train_copy_count})")
    print(f"测试集数量: {len(test_data)} (图片复制成功: {test_copy_count})")
    print(f"生成文件：\n  - {train_path}\n  - {test_path}")

if __name__ == "__main__":
    process_galaxy_data()
```

```python
import json
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
from matplotlib.gridspec import GridSpec

# 配置显示参数
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style="whitegrid", palette="muted")

# 类别定义
class_info = {
    'A': 'round elliptical',
    'B': 'in-between elliptical',
    'C': 'cigar-shaped elliptical',
    'D': 'edge-on',
    'E': 'Barred spirals',
    'F': 'Unbarred spirals',
    'G': 'Irregular',
    'H': 'merger'
}
labels = list(class_info.keys())
full_names = [f"{k}: {v}" for k, v in class_info.items()]

# 初始化统计容器
true_list, pred_list = [], []
confusion_counts = {true: {pred: 0 for pred in labels} for true in labels}

# 读取数据
file_path = r'/remote-home/cs_acmis_hby/code/FT-LLM/output/qwen_vl_finetune_01/v0-20250326-135312/checkpoint-3960/infer_result/20250327-014602.jsonl'
with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        data = json.loads(line.strip())
        pred = re.search(r'option\s+([A-Za-z])\s+is\s+selected', data['response']).group(1).upper()
        true = re.search(r'option\s+([A-Za-z])\s+is\s+selected', data['labels']).group(1).upper()
        true_list.append(true)
        pred_list.append(pred)
        confusion_counts[true][pred] += 1

# 转换为数值索引
label_idx = {k:i for i,k in enumerate(labels)}
y_true = [label_idx[t] for t in true_list]
y_pred = [label_idx[p] for p in pred_list]

# 计算综合指标
report = classification_report(y_true, y_pred, target_names=labels, output_dict=True)
cm = confusion_matrix(y_true, y_pred, labels=list(range(len(labels))))

# 指标数据框
metrics_df = pd.DataFrame({
    'Precision': [report[l]['precision'] for l in labels],
    'Recall': [report[l]['recall'] for l in labels],
    'F1-Score': [report[l]['f1-score'] for l in labels],
    'Support': [report[l]['support'] for l in labels]
}, index=full_names)

# 错误模式分析
error_matrix = cm.copy()
np.fill_diagonal(error_matrix, 0)
max_error = error_matrix.max()

# 错误模式表格
error_df = pd.DataFrame(
    [(full_names[i], full_names[j], error_matrix[i,j]) 
     for i in range(len(labels)) for j in range(len(labels)) if i != j],
    columns=['True Class', 'Predicted Class', 'Count']
).sort_values('Count', ascending=False).reset_index(drop=True)

# ========================
# 可视化函数
# ========================
def plot_confusion_matrix():
    """绘制混淆矩阵"""
    plt.figure(figsize=(18, 14), dpi=100)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=full_names, 
                yticklabels=full_names,
                annot_kws={"size": 14, "color":"white"},
                cbar_kws={"shrink":0.8})
    plt.title(f'Confusion Matrix (Accuracy: {report["accuracy"]:.2%})', 
             fontsize=18, pad=20)
    plt.xticks(rotation=45, fontsize=12)
    plt.yticks(rotation=0, fontsize=12)
    plt.tight_layout()
    plt.show()

def plot_class_distribution():
    """绘制类别分布图"""
    plt.figure(figsize=(16, 10), dpi=100)
    ax = sns.barplot(x='Support', y=metrics_df.index, 
                    palette='viridis',
                    data=metrics_df,
                    orient='h')
    plt.title('Class Distribution', fontsize=18, pad=15)
    plt.yticks(fontsize=12)
    plt.xlabel('Sample Count', fontsize=14)
    plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

def plot_radar_chart():
    """绘制雷达图"""
    plt.figure(figsize=(14, 14), dpi=100)
    ax = plt.subplot(111, polar=True)
    angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()
    angles += angles[:1]

    for metric, color in zip(['Precision', 'Recall', 'F1-Score'], 
                           ['#FF6B6B', '#4ECDC4', '#45B7D1']):
        values = metrics_df[metric].tolist()
        values += values[:1]
        ax.plot(angles, values, color=color, linewidth=3, 
               linestyle='solid', label=metric)
        ax.fill(angles, values, color=color, alpha=0.1)

    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(full_names, fontsize=12)
    ax.tick_params(axis='y', labelsize=10)
    ax.set_title('Class Metrics Radar Chart', y=1.08, fontsize=18)
    ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1), fontsize=12)
    plt.tight_layout()
    plt.show()

def plot_error_pattern():
    """绘制错误模式图"""
    plt.figure(figsize=(16, 12), dpi=100)
    sns.heatmap(error_matrix, cmap='Reds', 
                xticklabels=labels, 
                yticklabels=full_names,
                vmin=0, vmax=max_error,
                annot=True, fmt='d', 
                annot_kws={"size": 12, "color":"white"})
    plt.title('Error Pattern Analysis', fontsize=18, pad=15)
    plt.xticks(rotation=45, fontsize=12)
    plt.yticks(rotation=0, fontsize=12)
    plt.tight_layout()
    plt.show()

def plot_metrics_comparison():
    """绘制指标对比图"""
    plt.figure(figsize=(18, 8), dpi=100)
    metrics_df[['Precision', 'Recall', 'F1-Score']].plot(
        kind='bar', 
        color=['#FF9F43', '#55E6C1', '#2E86DE'],
        width=0.8,
        edgecolor='black',
        linewidth=1
    )
    plt.title('Performance Metrics Comparison', fontsize=18, pad=15)
    plt.xticks(rotation=45, ha='right', fontsize=12)
    plt.ylabel('Score', fontsize=14)
    plt.legend(loc='lower left', bbox_to_anchor=(1.02, 0), fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# ========================
# 主程序
# ========================
if __name__ == "__main__":
    # 依次显示图表
    plot_confusion_matrix()
    plot_class_distribution()
    plot_radar_chart()
    plot_error_pattern()
    plot_metrics_comparison()

    # 输出统计报告
    print("\n" + "="*80)
    print(f"{' Galaxy Classification Performance Report ':=^80}")
    print(metrics_df.to_string(float_format=lambda x: f"{x:.4f}"))
    
    print("\n" + "="*80)
    print(f"{' Top 5 Error Patterns ':=^80}")
    print(error_df.head(5).to_string(index=False))
```

```python
import os
import matplotlib.pyplot as plt
from PIL import Image

def draw_galaxyID(n, df):
    # 确保抽取唯一样本
    return df.sample(n=min(n, len(df)), replace=False).galaxyID.values

def plot_samples(df, class_name, dir_image, figsize=(12, 4)):
    # 一次性抽取12个唯一ID
    sampled_ids = draw_galaxyID(12, df)
    if len(sampled_ids) < 12:
        print(f"Warning: 数据集 '{class_name}' 样本不足12个，仅显示 {len(sampled_ids)} 张。")
    
    # 动态调整子图布局
    n_rows = 2
    n_cols = 6
    fig, axes = plt.subplots(n_rows, n_cols, figsize=figsize)
    plt.subplots_adjust(hspace=0.2, wspace=0.2)
    
    for i, ax in enumerate(axes.flat):
        if i >= len(sampled_ids):
            ax.axis('off')
            continue
        
        imgID = sampled_ids[i]
        try:
            img = Image.open(os.path.join(dir_image, f"{imgID}.jpg"))
            ax.imshow(img)
            ax.text(0.5, 0.95, f'Class: {class_name}\n{imgID}', 
                    transform=ax.transAxes, color='white', ha='center', fontsize=8)
        except FileNotFoundError:
            ax.text(0.5, 0.5, "Image Not Found", ha='center', va='center')
        
        ax.axis('off')
    
    plt.suptitle(f"Galaxy Samples - {class_name}", y=0.98)
    plt.show()

# 调用示例
dir_image = "/remote-home/cs_acmis_hby/AstroData/images_gz2/images"  # 明确图像路径
plot_samples(df_class0, "Elliptical", dir_image)
plot_samples(df_class1, "Spiral", dir_image)
```
